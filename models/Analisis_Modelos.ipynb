## 2. Procesamiento y Transformación a DataFrame
Aplanamos el JSON para convertirlo en un formato tabular manejable por Pandas.

```python
rows = []
for version, sub_dict in data.items():
    for subversion, metrics in sub_dict.items():
        rows.append({
            'Version': version,
            'Subversion': subversion.capitalize(),
            'F1-Score': metrics['f1'],
            'Accuracy': metrics['accuracy'],
            'Modelo': metrics['modelo']
        })

df = pd.DataFrame(rows)
display(df.head())
```

## 3. Creación de la Tabla Comparativa
Pivotamos la tabla para tener las 3 secciones (Original, Undersampling, Undersampling + VAE) en columnas y las subversiones en filas.

```python
tabla_f1 = df.pivot(index='Subversion', columns='Version', values='F1-Score')

# Reordenar para la visualización
tabla_f1 = tabla_f1[['Original', 'Undersampling', 'Undersampling + VAE']]
tabla_f1 = tabla_f1.reindex(['Codigos', 'Descripciones', 'Combinado'])

print("=== Tabla Comparativa de F1-Score ===")
display(tabla_f1)
```

## 4. Visualización de los Resultados
Generamos una gráfica de barras agrupadas para comparar visualmente cuál combinación da los mejores resultados.

```python
plt.figure(figsize=(10, 6))
sns.barplot(data=df, x='Subversion', y='F1-Score', hue='Version', 
            palette=['#4C72B0', '#DD8452', '#55A868'], 
            order=['Codigos', 'Descripciones', 'Combinado'])

plt.title('Comparativa de F1-Score por Dataset y Preprocesamiento', fontsize=14, pad=15)
plt.xlabel('Tipo de Información', fontsize=12)
plt.ylabel('F1-Score', fontsize=12)
plt.ylim(0, 1.0) # Para que la escala vaya de 0 a 1
plt.legend(title='Versión del Dataset', bbox_to_anchor=(1.05, 1), loc='upper left')

# Añadir los valores numéricos sobre cada barra
ax = plt.gca()
for p in ax.patches:
    ax.annotate(f'{p.get_height():.3f}', (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='center', fontsize=10, color='black', xytext=(0, 5),
                textcoords='offset points')

plt.tight_layout()
plt.show()
```
