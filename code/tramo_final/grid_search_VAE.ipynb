```python
import pandas as pd
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import classification_report, confusion_matrix
```

```python
# 1. Cargar el dataset híbrido (Reales + VAE)
df = pd.read_csv('embeddings_entrenamiento_vae.csv')

# 2. Preparar X e y
X = df.drop(['DIAG PSQ', 'Origen'], axis=1)
y = df['DIAG PSQ']
```

```python
# 3. Definir la Grid Search para el clasificador final
# Aquí probamos diferentes "cerebros" para leer esos embeddings
param_grid = {
    'hidden_layer_sizes': [(256,), (512, 256), (256, 128, 64), (512, 256, 128), (256, 128, 64, 32)],
    'alpha': [0.0001, 0.001, 0.01], # Regularización para no memorizar los sintéticos
    'learning_rate_init': [0.001, 0.0001, 0.01], # Velocidad de aprendizaje
    'max_iter': [500, 1000, 1500, 300] # Iteraciones para asegurar convergencia
}
```

```python
mlp = MLPClassifier(random_state=42)
grid = GridSearchCV(mlp, param_grid, cv=5, scoring='f1_macro', verbose=2)

print("Iniciando búsqueda del mejor clasificador sobre datos aumentados...")
grid.fit(X, y)

print(f"Mejor configuración: {grid.best_params_}")
```
